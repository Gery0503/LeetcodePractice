### Description

Given a valid (IPv4) IP `address`, return a defanged version of that IP address.

A *defanged IP address* replaces every period `"."` with `"[.]"`.

**Example 1:**
```
Input: address = "1.1.1.1"
Output: "1[.]1[.]1[.]1"
```

**Example 2:**
```
Input: address = "255.100.50.0"
Output: "255[.]100[.]50[.]0"
```

**Constraints:**
- The given `address` is a valid IPv4 address.

````python
# Concatenation approach: Time: O(N**2)
"""
Not a good one.
Because strings are immutable in Python, they cannot be changed in place. 
Every time you do encoded += char, Python has to allocate a entirely new chunk of memory, 
copy the old string into it, and then add the new character.
"""
def defangIPaddr(address: str) -> str:
    encoded = ''
    for char in address:
        if char == '.':
            char = '[.]'
        encoded += char
    return encoded
````

````python
# Use list: O(N)
# return ''.join('[.]' if c == '.' else c for c in address)
def defangIPaddr(address: str) -> str:
    encoded = []
    for char in address:
        if char == '.':
            char = '[.]'
        encoded.append(char)
    return ''.join(encoded)

if __name__ == '__main__':
    # Test Case 1
    address1 = "1.1.1.1"
    result1 = defangIPaddr(address1)
    print(f"Test 1 Input: address = \"{address1}\"")
    print(f"Test 1 Output: \"{result1}\"") # Expected: "1[.]1[.]1[.]1"

    # Test Case 2
    address2 = "255.100.50.0"
    result2 = defangIPaddr(address2)
    print(f"\nTest 2 Input: address = \"{address2}\"")
    print(f"Test 2 Output: \"{result2}\"") # Expected: "255[.]100[.]50[.]0"
````
