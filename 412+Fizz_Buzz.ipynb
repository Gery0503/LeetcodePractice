# 412. Fizz Buzz

[Link to Problem](https://leetcode.com/problems/fizz-buzz/)

### Description

Given an integer `n`, return a string array `answer` (1-indexed) where:

- `answer[i] == "FizzBuzz"` if `i` is divisible by `3` and `5`.
- `answer[i] == "Fizz"` if `i` is divisible by `3`.
- `answer[i] == "Buzz"` if `i` is divisible by `5`.
- `answer[i] == i` (as a string) if none of the above conditions are true.

**Example 1:**
```
Input: n = 3
Output: ["1","2","Fizz"]
```

**Example 2:**
```
Input: n = 5
Output: ["1","2","Fizz","4","Buzz"]
```

**Example 3:**
```
Input: n = 15
Output: ["1","2","Fizz","4","Buzz","Fizz","7","8","Fizz","Buzz","11","Fizz","13","14","FizzBuzz"]
```

**Constraints:**
- `1 <= n <= 10^4`

````python
# Gemini version
def encode_scalable(n: int) -> str:
    res = ""
    if n % 3 == 0:
        res += "Fizz"
    if n % 5 == 0:
        res += "Buzz"
        
    return res if res else str(n)
````

````python
# History version

def encode(n: int) -> str:
    if n % 15 == 0:
        return 'FizzBuzz'
    if n % 3 == 0:
        return 'Fizz'
    if n % 5 == 0:
        return 'Buzz'
    return str(n)

def fizzBuzz(n: int) -> List[str]:
    return [encode(i) for i in range(1, n+1)]

if __name__ == '__main__':
    # Test Case 1
    n1 = 3
    result1 = fizzBuzz(n1)
    print(f"Test 1 Input: n = {n1}")
    print(f"Test 1 Output: {result1}") # Expected: ["1","2","Fizz"]

    # Test Case 2
    n2 = 5
    result2 = fizzBuzz(n2)
    print(f"\nTest 2 Input: n = {n2}")
    print(f"Test 2 Output: {result2}") # Expected: ["1","2","Fizz","4","Buzz"]

    # Test Case 3
    n3 = 15
    result3 = fizzBuzz(n3)
    print(f"\nTest 3 Input: n = {n3}")
    print(f"Test 3 Output: {result3}") # Expected: ["1","2","Fizz","4","Buzz","Fizz","7","8","Fizz","Buzz","11","Fizz","13","14","FizzBuzz"]
````
