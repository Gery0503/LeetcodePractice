# 1051. Height Checker

[Link to Problem](https://leetcode.com/problems/height-checker/)

### Description

A school is trying to take an annual photo of all the students. The students are asked to stand in a single file line in non-decreasing order by height. Let this ordering be represented by the integer array `expected` where `expected[i]` is the expected height of the `i`th student in line.

You are given an integer array `heights` representing the current order that the students are standing in. Each `heights[i]` is the height of the `i`th student in line (0-indexed).

Return _the number of indices where_ `heights[i] != expected[i]`.

**Example 1:**
```
Input: heights = [1,1,4,2,1,3]
Output: 3
Explanation: 
heights: [1,1,4,2,1,3]
expected: [1,1,1,2,3,4]
Indices 2, 4, and 5 do not match.
```

**Example 2:**
```
Input: heights = [5,1,2,3,4]
Output: 5
Explanation:
heights: [5,1,2,3,4]
expected: [1,2,3,4,5]
All indices do not match.
```

**Example 3:**
```
Input: heights = [1,2,3,4,5]
Output: 0
Explanation:
heights: [1,2,3,4,5]
expected: [1,2,3,4,5]
All indices match.
```

**Constraints:**
- `1 <= heights.length <= 100`
- `1 <= heights[i] <= 100`

## My Intuition

_Write down your thoughts, ideas, and insights here._

- **Observations:**
  1. Given the current heights. I could sort first to get expected heights and then compare to current heights. 
- **Edge cases:**
  - An already sorted array (e.g., [1, 2, 3, 4, 5]) → Expected output: 0
  - An array with duplicate elements / all identical elements (e.g., [2, 2, 2, 2]) → Expected output: 0
  - An array of minimum length (1) → Expected output: 0
- **Expected approach and complexity:**
  1. Time:  O(n+k where k=100)
  2. Space: O(n)

````python
# Counting sort (best solution)
from typing import List

def heightChecker(heights: List[int]) -> int:
    # Count frequencies of heights (since values are between 1 and 100)
    count = [0] * 101
    for h in heights:
        count[h] += 1
        
    mismatches = 0
    current_height = 1
    for h in heights:
        # Advance to the next height that actually exists in count
        while count[current_height] == 0:
            current_height += 1
        if h != current_height:
            mismatches += 1
        count[current_height] -= 1
        
    return mismatches

if __name__ == '__main__':
    # Test Case 1
    heights1 = [1, 1, 4, 2, 1, 3]
    result1 = heightChecker(heights1)
    print(f"Test 1 Input: heights = {heights1}")
    print(f"Test 1 Output: {result1}") # Expected: 3

    # Test Case 2
    heights2 = [5, 1, 2, 3, 4]
    result2 = heightChecker(heights2)
    print(f"\nTest 2 Input: heights = {heights2}")
    print(f"Test 2 Output: {result2}") # Expected: 5

    # Test Case 3
    heights3 = [1, 2, 3, 4, 5]
    result3 = heightChecker(heights3)
    print(f"\nTest 3 Input: heights = {heights3}")
    print(f"Test 3 Output: {result3}") # Expected: 0
````

````python
# Pyhonic solution O(nlogn)
from typing import List

def heightChecker(heights: List[int]) -> int:
    return sum(h1 != h2 for h1, h2 in zip(heights, sorted(heights)))

if __name__ == '__main__':
    # Test Case 1
    heights1 = [1, 1, 4, 2, 1, 3]
    result1 = heightChecker(heights1)
    print(f"Test 1 Input: heights = {heights1}")
    print(f"Test 1 Output: {result1}") # Expected: 3

    # Test Case 2
    heights2 = [5, 1, 2, 3, 4]
    result2 = heightChecker(heights2)
    print(f"\nTest 2 Input: heights = {heights2}")
    print(f"Test 2 Output: {result2}") # Expected: 5

    # Test Case 3
    heights3 = [1, 2, 3, 4, 5]
    result3 = heightChecker(heights3)
    print(f"\nTest 3 Input: heights = {heights3}")
    print(f"Test 3 Output: {result3}") # Expected: 0
````
