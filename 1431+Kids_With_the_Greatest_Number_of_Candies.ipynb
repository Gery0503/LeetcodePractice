# 1431. Kids With the Greatest Number of Candies

[Link to Problem](https://leetcode.com/problems/kids-with-the-greatest-number-of-candies/)

### Description

There are `n` kids with candies. You are given an integer array `candies`, where each `candies[i]` represents the number of candies the `i`th kid has, and an integer `extraCandies`, denoting the number of extra candies that you have.

Return _a boolean array_ `result` _of length_ `n`, _where_ `result[i]` _is_ `true` _if, after giving the_ `i`th _kid all the_ `extraCandies`, _they will have the **greatest** number of candies among all the kids, or_ `false` _otherwise_.

Note that **multiple** kids can have the **greatest** number of candies.

**Example 1:**
```
Input: candies = [2,3,5,1,3], extraCandies = 3
Output: [true,true,true,false,true] 
Explanation: If you give all extraCandies to:
- Kid 1, they will have 2 + 3 = 5 candies, which is the greatest among the kids.
- Kid 2, they will have 3 + 3 = 6 candies, which is the greatest among the kids.
- Kid 3, they will have 5 + 3 = 8 candies, which is the greatest among the kids.
- Kid 4, they will have 1 + 3 = 4 candies, which is not the greatest among the kids.
- Kid 5, they will have 3 + 3 = 6 candies, which is the greatest among the kids.
```

**Example 2:**
```
Input: candies = [4,2,1,1,2], extraCandies = 1
Output: [true,false,false,false,false] 
Explanation: There is only 1 extra candy. 
Therefore, only kid 1 will have the greatest number of candies among the kids regardless of who takes the extra candy.
```

**Example 3:**
```
Input: candies = [12,1,12], extraCandies = 10
Output: [true,false,true]
```

**Constraints:**
- `n == candies.length`
- `2 <= n <= 100`
- `1 <= candies[i] <= 100`
- `1 <= extraCandies <= 50`

## My Intuition

_Write down your thoughts, ideas, and insights here._

- **Observations:**
  1. For each kid, we need to ask: "If I give this kid all the extra candies, will they have greater than or equal to the person who currently has the most?"
  2. Instead of re-calculating who has the most candies on every single turn, we can just find the maximum value in the array once at the very beginning.
- **Edge cases:**
  - Multiple maximums: The problem allows multiple kids to share the "greatest" number of candies, so our condition is strictly >= max_candies.
- **Expected approach and complexity:**
  1. Time: O(N)
  2. Space: O(1)

````python
from typing import List

# classic "Two-Pass" array problem
def kidsWithCandies(candies: List[int], extraCandies: int) -> List[bool]:
    max_candies = max(candies)
    threshold = max_candies - extraCandies
    return [num >= threshold for num in candies]

if __name__ == '__main__':
    # Test Case 1
    candies1 = [2, 3, 5, 1, 3]
    extraCandies1 = 3
    result1 = kidsWithCandies(candies1, extraCandies1)
    print(f"Test 1 Input: candies = {candies1}, extraCandies = {extraCandies1}")
    print(f"Test 1 Output: {result1}") # Expected: [True, True, True, False, True]

    # Test Case 2
    candies2 = [4, 2, 1, 1, 2]
    extraCandies2 = 1
    result2 = kidsWithCandies(candies2, extraCandies2)
    print(f"\nTest 2 Input: candies = {candies2}, extraCandies = {extraCandies2}")
    print(f"Test 2 Output: {result2}") # Expected: [True, False, False, False, False]

    # Test Case 3
    candies3 = [12, 1, 12]
    extraCandies3 = 10
    result3 = kidsWithCandies(candies3, extraCandies3)
    print(f"\nTest 3 Input: candies = {candies3}, extraCandies = {extraCandies3}")
    print(f"Test 3 Output: {result3}") # Expected: [True, False, True]
````
