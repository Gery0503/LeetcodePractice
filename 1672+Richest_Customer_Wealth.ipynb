# 1672. Richest Customer Wealth

[Link to Problem](https://leetcode.com/problems/richest-customer-wealth/)

### Description

You are given an `m x n` integer grid `accounts` where `accounts[i][j]` is the amount of money the `i`th customer has in the `j`th bank. Return _the **wealth** that the richest customer has_.

A customer's **wealth** is the amount of money they have in all their bank accounts. The richest customer is the customer that has the maximum **wealth**.

**Example 1:**
```
Input: accounts = [[1,2,3],[3,2,1]]
Output: 6
Explanation:
1st customer has wealth = 1 + 2 + 3 = 6
2nd customer has wealth = 3 + 2 + 1 = 6
Both customers are considered the richest with a wealth of 6 each, so return 6.
```

**Example 2:**
```
Input: accounts = [[1,5],[7,3],[3,5]]
Output: 10
Explanation: 
1st customer has wealth = 6
2nd customer has wealth = 10 
3rd customer has wealth = 8
The 2nd customer is the richest with a wealth of 10.
```

**Example 3:**
```
Input: accounts = [[2,8,7],[7,1,3],[1,9,5]]
Output: 17
```

**Constraints:**
- `m == accounts.length`
- `n == accounts[i].length`
- `1 <= m, n <= 50`
- `1 <= accounts[i][j] <= 100`

## My Intuition

_Write down your thoughts, ideas, and insights here._

- **Observations:**
  1. Simulation or Exhaustive Search
  2. The objective is simple: sum up the numbers in each row independently, and find the maximum resulting sum.
- **Edge cases:**
  - 
- **Expected approach and complexity:**
  1. Time:  O(m*n)
  2. Space: O(1)

````python
from typing import List

def maximumWealth(accounts: List[List[int]]) -> int:
    wealth = 0
    for account in accounts:
        wealth = max(sum(account), wealth)
    return wealth
    
if __name__ == '__main__':
    # Test Case 1
    accounts1 = [[1, 2, 3], [3, 2, 1]]
    result1 = maximumWealth(accounts1)
    print(f"Test 1 Input: accounts = {accounts1}")
    print(f"Test 1 Output: {result1}") # Expected: 6

    # Test Case 2
    accounts2 = [[1, 5], [7, 3], [3, 5]]
    result2 = maximumWealth(accounts2)
    print(f"\nTest 2 Input: accounts = {accounts2}")
    print(f"Test 2 Output: {result2}") # Expected: 10

    # Test Case 3
    accounts3 = [[2, 8, 7], [7, 1, 3], [1, 9, 5]]
    result3 = maximumWealth(accounts3)
    print(f"\nTest 3 Input: accounts = {accounts3}")
    print(f"Test 3 Output: {result3}") # Expected: 17
````
