# 263. Ugly Number

[Link to Problem](https://leetcode.com/problems/ugly-number/)

### Description

An **ugly number** is a positive integer whose prime factors are limited to `2`, `3`, and `5`.

Given an integer `n`, return `true` if `n` is an **ugly number**.

**Example 1:**
```
Input: n = 6
Output: true
Explanation: 6 = 2 x 3
```

**Example 2:**
```
Input: n = 1
Output: true
Explanation: 1 has no prime factors, therefore all of its prime factors are limited to 2, 3, and 5.
```

**Example 3:**
```
Input: n = 14
Output: false
Explanation: 14 is not ugly since it includes the prime factor 7.
```

**Constraints:**
- `-2^31 <= n <= 2^31 - 1`

## My Intuition

_Write down your thoughts, ideas, and insights here._

- **Observations:**
  1. divide 2 until the remainder is not 0
  2. divide 3 until the remainder is not 0
  3. divide 5 until the remainder is not 0
  4. if remainder is 1 return True else False
- **Edge cases:**
  - `n = 0`: Must be handled immediately. If not caught, evaluating `while 0 % 2 == 0:` would result in an infinite loop.
  - `n < 0`: The problem defines ugly numbers strictly as positive integers, so all negative numbers automatically return `False`.
  - `n = 1`: A base case that must return `True`. Since 1 has no prime factors, it trivially satisfies the condition that its prime factors are limited to 2, 3, and 5.
- **Expected approach and complexity:**
  1. Time:  O(logn)
  2. Space: O(1)

````python
def isUgly(n: int) -> bool:
    if n <= 0:
        return False
        
    for p in [2, 3, 5]:
        while n % p == 0:
            n //= p
            
    return n == 1

if __name__ == '__main__':
    # Test Case 1
    n1 = 6
    result1 = isUgly(n1)
    print(f"Test 1 Input: n = {n1}")
    print(f"Test 1 Output: {result1}") # Expected: True

    # Test Case 2
    n2 = 1
    result2 = isUgly(n2)
    print(f"\nTest 2 Input: n = {n2}")
    print(f"Test 2 Output: {result2}") # Expected: True

    # Test Case 3
    n3 = 14
    result3 = isUgly(n3)
    print(f"\nTest 3 Input: n = {n3}")
    print(f"Test 3 Output: {result3}") # Expected: False

    # Test Case 4
    n4 = -2147483648
    result4 = isUgly(n4)
    print(f"\nTest 4 Input: n = {n4}")
    print(f"Test 4 Output: {result4}") # Expected: False
````
